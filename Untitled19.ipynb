```python
# ============================================================
# DATA SCIENCE - TASK 3
# EXPLORATORY DATA ANALYSIS (EDA) PROJECT
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("=" * 70)
print("EXPLORATORY DATA ANALYSIS (EDA)")
print("=" * 70)


# ============================================================
# 1. CREATE DATASET
# ============================================================

data = {
    "Customer_ID": [
        "C001","C002","C003","C004","C005",
        "C006","C007","C008","C009","C010",
        "C011","C012","C013","C014","C015",
        "C016","C017","C018","C019","C020",
        "C021","C022","C023","C024","C025"
    ],

    "Age": [
        22,45,31,52,27,
        39,24,48,35,29,
        55,26,42,33,51,
        23,37,46,30,58,
        28,41,34,49,25
    ],

    "Annual_Income": [
        28000,85000,52000,92000,35000,
        68000,31000,105000,59000,45000,
        115000,33000,78000,61000,98000,
        27000,72000,88000,49000,120000,
        40000,76000,55000,95000,32000
    ],

    "Purchase_Frequency": [
        8,22,15,25,10,
        19,7,28,17,13,
        30,9,21,16,26,
        6,20,24,14,32,
        11,18,12,27,8
    ],

    "Total_Spend": [
        12000,78000,42000,95000,18000,
        62000,10000,110000,50000,30000,
        135000,15000,72000,46000,88000,
        9000,65000,92000,38000,145000,
        22000,70000,36000,105000,14000
    ],

    "City": [
        "Bangalore","Mysore","Chennai","Bangalore","Mysore",
        "Chennai","Bangalore","Mysore","Chennai","Bangalore",
        "Bangalore","Chennai","Mysore","Bangalore","Chennai",
        "Mysore","Bangalore","Chennai","Mysore","Bangalore",
        "Chennai","Mysore","Bangalore","Chennai","Mysore"
    ],

    "Satisfaction_Score": [
        6,8,7,9,5,
        8,4,9,7,6,
        10,5,8,7,9,
        3,8,9,5,10,
        6,8,7,9,5
    ]
}

df = pd.DataFrame(data)


# ============================================================
# 2. DISPLAY DATASET
# ============================================================

print("\nDATASET")
print("-" * 70)

display(df)


# ============================================================
# 3. BASIC DATA INFORMATION
# ============================================================

print("\nDATASET INFORMATION")
print("-" * 70)

print("Number of Rows    :", df.shape[0])
print("Number of Columns :", df.shape[1])

print("\nData Types:")
print(df.dtypes)


# ============================================================
# 4. CHECK MISSING VALUES
# ============================================================

print("\nMISSING VALUES")
print("-" * 70)

print(df.isnull().sum())


# ============================================================
# 5. CHECK DUPLICATES
# ============================================================

print("\nDUPLICATES")
print("-" * 70)

print("Duplicate records:", df.duplicated().sum())


# ============================================================
# 6. STATISTICAL SUMMARY
# ============================================================

print("\nSTATISTICAL SUMMARY")
print("-" * 70)

display(df.describe())


# ============================================================
# 7. MEDIAN, MODE AND STANDARD DEVIATION
# ============================================================

numeric_columns = [
    "Age",
    "Annual_Income",
    "Purchase_Frequency",
    "Total_Spend",
    "Satisfaction_Score"
]

summary = pd.DataFrame({
    "Mean": df[numeric_columns].mean(),
    "Median": df[numeric_columns].median(),
    "Standard_Deviation": df[numeric_columns].std(),
    "Minimum": df[numeric_columns].min(),
    "Maximum": df[numeric_columns].max()
})

print("\nDETAILED STATISTICAL SUMMARY")
print("-" * 70)

display(summary.round(2))


# ============================================================
# 8. CITY-WISE ANALYSIS
# ============================================================

city_analysis = df.groupby("City").agg(
    Customers=("Customer_ID", "count"),
    Average_Income=("Annual_Income", "mean"),
    Average_Spend=("Total_Spend", "mean"),
    Total_Spend=("Total_Spend", "sum"),
    Average_Satisfaction=("Satisfaction_Score", "mean")
).reset_index()

print("\nCITY-WISE ANALYSIS")
print("-" * 70)

display(city_analysis.round(2))


# ============================================================
# 9. AGE DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    df["Age"],
    bins=8,
    kde=True
)

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()


# ============================================================
# 10. INCOME DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    df["Annual_Income"],
    bins=8,
    kde=True
)

plt.title("Annual Income Distribution")
plt.xlabel("Annual Income")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()


# ============================================================
# 11. PURCHASE FREQUENCY DISTRIBUTION
# ============================================================

plt.figure(figsize=(9, 5))

sns.histplot(
    df["Purchase_Frequency"],
    bins=8,
    kde=True
)

plt.title("Purchase Frequency Distribution")
plt.xlabel("Purchase Frequency")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()


# ============================================================
# 12. CITY-WISE CUSTOMER COUNT
# ============================================================

plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="City"
)

plt.title("Customer Count by City")
plt.xlabel("City")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()


# ============================================================
# 13. CITY-WISE TOTAL SPENDING
# ============================================================

plt.figure(figsize=(9, 5))

sns.barplot(
    data=city_analysis,
    x="City",
    y="Total_Spend"
)

plt.title("Total Customer Spending by City")
plt.xlabel("City")
plt.ylabel("Total Spending")

plt.tight_layout()
plt.show()


# ============================================================
# 14. INCOME VS TOTAL SPEND
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Annual_Income",
    y="Total_Spend",
    hue="City",
    s=120
)

plt.title("Annual Income vs Total Spend")
plt.xlabel("Annual Income")
plt.ylabel("Total Spend")

plt.tight_layout()
plt.show()


# ============================================================
# 15. PURCHASE FREQUENCY VS TOTAL SPEND
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Purchase_Frequency",
    y="Total_Spend",
    hue="City",
    s=120
)

plt.title("Purchase Frequency vs Total Spend")
plt.xlabel("Purchase Frequency")
plt.ylabel("Total Spend")

plt.tight_layout()
plt.show()


# ============================================================
# 16. AGE VS SATISFACTION
# ============================================================

plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x="Age",
    y="Satisfaction_Score",
    hue="City",
    s=120
)

plt.title("Age vs Satisfaction Score")
plt.xlabel("Age")
plt.ylabel("Satisfaction Score")

plt.tight_layout()
plt.show()


# ============================================================
# 17. CORRELATION MATRIX
# ============================================================

correlation = df[numeric_columns].corr()

print("\nCORRELATION MATRIX")
print("-" * 70)

display(correlation.round(2))


# ============================================================
# 18. CORRELATION HEATMAP
# ============================================================

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()


# ============================================================
# 19. BOX PLOT
# ============================================================

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df[numeric_columns]
)

plt.title("Distribution and Outlier Analysis")
plt.xticks(rotation=20)

plt.tight_layout()
plt.show()


# ============================================================
# 20. KEY INFLUENCING FACTORS
# ============================================================

spend_correlations = correlation["Total_Spend"].drop(
    "Total_Spend"
).sort_values(
    ascending=False
)

print("\nFACTORS RELATED TO TOTAL SPEND")
print("-" * 70)

display(spend_correlations.round(3))


# ============================================================
# 21. FIND STRONGEST CORRELATION
# ============================================================

strongest_factor = spend_correlations.abs().idxmax()

strongest_value = spend_correlations[
    strongest_factor
]

print("\nStrongest relationship with Total Spend:")
print(strongest_factor)

print(
    "Correlation:",
    round(strongest_value, 3)
)


# ============================================================
# 22. HIGHEST SPENDING CITY
# ============================================================

highest_spending_city = city_analysis.loc[
    city_analysis["Total_Spend"].idxmax(),
    "City"
]

highest_spending_amount = city_analysis[
    "Total_Spend"
].max()

print("\nHIGHEST SPENDING CITY")
print("-" * 70)

print("City:", highest_spending_city)

print(
    "Total Spend: ₹",
    f"{highest_spending_amount:,.2f}"
)


# ============================================================
# 23. HIGHEST SATISFACTION CITY
# ============================================================

highest_satisfaction_city = city_analysis.loc[
    city_analysis["Average_Satisfaction"].idxmax(),
    "City"
]

print("\nHIGHEST AVERAGE SATISFACTION CITY")
print("-" * 70)

print(highest_satisfaction_city)


# ============================================================
# 24. TOP SPENDING CUSTOMERS
# ============================================================

top_customers = df.sort_values(
    by="Total_Spend",
    ascending=False
).head(5)

print("\nTOP 5 CUSTOMERS BY SPENDING")
print("-" * 70)

display(
    top_customers[
        [
            "Customer_ID",
            "City",
            "Annual_Income",
            "Purchase_Frequency",
            "Total_Spend",
            "Satisfaction_Score"
        ]
    ]
)


# ============================================================
# 25. EDA DASHBOARD
# ============================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(14, 10)
)

sns.histplot(
    df["Age"],
    bins=8,
    kde=True,
    ax=axes[0, 0]
)

axes[0, 0].set_title("Age Distribution")


sns.barplot(
    data=city_analysis,
    x="City",
    y="Total_Spend",
    ax=axes[0, 1]
)

axes[0, 1].set_title("Total Spending by City")


sns.scatterplot(
    data=df,
    x="Annual_Income",
    y="Total_Spend",
    ax=axes[1, 0]
)

axes[1, 0].set_title("Income vs Total Spend")


sns.boxplot(
    data=df,
    x="City",
    y="Satisfaction_Score",
    ax=axes[1, 1]
)

axes[1, 1].set_title("Satisfaction by City")


plt.suptitle(
    "Exploratory Data Analysis Dashboard",
    fontsize=18
)

plt.tight_layout()
plt.show()


# ============================================================
# 26. STRUCTURED EDA REPORT
# ============================================================

print("\n" + "=" * 70)
print("STRUCTURED EDA REPORT")
print("=" * 70)

print("\n1. DATASET OVERVIEW")
print(
    f"The dataset contains {df.shape[0]} customers "
    f"and {df.shape[1]} variables."
)

print("\n2. STATISTICAL FINDINGS")
print(
    f"Average customer age: {df['Age'].mean():.2f} years"
)

print(
    f"Average annual income: ₹{df['Annual_Income'].mean():,.2f}"
)

print(
    f"Average total spending: ₹{df['Total_Spend'].mean():,.2f}"
)

print(
    f"Average purchase frequency: "
    f"{df['Purchase_Frequency'].mean():.2f}"
)

print("\n3. CORRELATION FINDINGS")
print(
    f"The strongest relationship with Total Spend "
    f"is observed for {strongest_factor}, "
    f"with correlation {strongest_value:.3f}."
)

print("\n4. CITY ANALYSIS")
print(
    f"{highest_spending_city} has the highest total "
    f"customer spending in this dataset."
)

print("\n5. SATISFACTION ANALYSIS")
print(
    f"{highest_satisfaction_city} has the highest "
    f"average satisfaction score."
)

print("\n6. DATA STORY")
print(
    "The EDA shows how customer income, purchasing "
    "frequency, age, city and satisfaction are related "
    "to customer spending. Visualizations and correlation "
    "analysis help identify important patterns and "
    "relationships in the dataset."
)


# ============================================================
# 27. SAVE EDA RESULTS
# ============================================================

df.to_csv(
    "eda_customer_dataset.csv",
    index=False
)

city_analysis.to_csv(
    "eda_city_analysis.csv",
    index=False
)

summary.to_csv(
    "eda_statistical_summary.csv"
)

print("\nFILES CREATED")
print("-" * 70)

print("1. eda_customer_dataset.csv")
print("2. eda_city_analysis.csv")
print("3. eda_statistical_summary.csv")


# ============================================================
# FINAL RESULT
# ============================================================

print("\n" + "=" * 70)
print("EDA PROJECT COMPLETED SUCCESSFULLY")
print("=" * 70)

print("""
Completed Components:

✓ Dataset exploration
✓ Statistical summaries
✓ Mean, Median and Standard Deviation
✓ Missing value checking
✓ Duplicate checking
✓ Distribution analysis
✓ City-wise analysis
✓ Spending analysis
✓ Correlation analysis
✓ Correlation heatmap
✓ Scatter plots
✓ Box plots
✓ Data visualization
✓ Key influencing factors
✓ Pattern identification
✓ Trend analysis
✓ Structured EDA report
✓ Data storytelling
✓ CSV result export
""")
```
